```python
def printlist(gatecs):
    for i in gatecs:
        if isinstance(i, list):              # funaction to check if "i" is a list also or not                        
                for j in i:                  #  "j" iterates over "i"
                    if isinstance(j,list):
                            for k in j:      #  "k" iterates over "j"
                                print(k)           

                    else:
                            print(j)         # if "j" is not a list print "j"
                    
        else:
                print(i)

gatecs = ['ADA' , ['CLR', 'SS' ,[12]], 'DM', ['RS','SH']]      # 3D list 
printlist(gatecs)
            
```

```python
def matrixpower(A,n):
    #B = A
    B =[i[:] for i in A]     #it create copy of that list
    #C = A
    C = [i[:] for i in  A]
    for i in range(n):
        C[0][0] = A[0][0] * B[0][0] + A[0][1] * B[1][0]
        C[0][1] = A[0][0] * B[0][1] + A[0][1] * B[1][1]
        C[1][0] = A[1][0] * B[0][0] + A[1][1] * B[1][0]
        C[1][1] = A[1][0] * B[0][1] + A[1][1] * B[1][1]
        
        #B = C
        B = [i[:] for i in C]
    return C


A = [[1,1],[1,0]] #jupyter prints last value without using print statement
print(matrixpower(A,2))

for n in range(10):
    result = matrixpower(A, n)
    ##print(result)
    print(n,result[0][0])  #This will be here fibonacci series when we multiply A*n times
                           # its basically the values of first elements of different matrices
```
