```python
with open(r"C:\Users\ibrah\OneDrive\Desktop\ibrahim.txt","r") as d:
  data=d.readlines()   
  
  print(data)
  
 
  length = len(data)
   
  for i in range(length):
    data[i] = data[i].lower()  # converting every line lowercase
  print(data[0])
  

  for i in range(length):
    data[i] = data[i].replace('\n'," ")  # remove new line character


unchars = ['\n','.',',']
for i in range(length):             # remove unwanted characters
  for j in unchars:
    data[i] = data[i].replace(j," ")

singlestring = " "                  # make one big string
for i in range(length):
  singlestring += data[i] + " "     # combine all lines ,all lines merged into one string


print(singlestring)
words = singlestring.split()   #split into words as a list
print(words)
print(len(words))       # count total words


dict = {}   


for i in words:
  if i in dict:
    dict[i] += 1   #if word already exists increase count
  else:
    dict[i] = 1  # if nrew appers create entry with count 1
for i in dict:
  print(i,':',dict[i])
  
```
